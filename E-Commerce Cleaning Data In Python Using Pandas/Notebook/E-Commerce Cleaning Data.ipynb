```python
# drop of duplicates in the df
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
df
```

```python
df=df.reset_index().drop(columns="index")
df
```

```python
# checking the null or missing vlues in the df
df.isna().sum()
```

```python
df.dropna(inplace=True)
df
```

```python
# the first latter to capitals
df["BrandName"]=df["BrandName"].apply(lambda x: x.capitalize())
df["Details"]=df["Details"].apply(lambda x: x.capitalize())
df
```

```python
# Change  "Nan" to "NA VALUED" values from  the "Sizes" in the  df
df["Sizes"]=df["Sizes"].replace("Nan", "NA VALUED")
df
```

```python
# ckecking the not "NA VALUES" fron the Sizes in the df
df=df[df["Sizes"] !="NA VALUED"]
df
```

```python
df.reset_index().drop(columns="index")
```

```python
# Remove nan values from the Discount in the df
df["Discount"]=df["Discount"].replace("Nan", "No Discount")
df
```

```python
# the save cleaning dataset
df.to_csv("Cleaned_Dataset.csv", index=False)
```
